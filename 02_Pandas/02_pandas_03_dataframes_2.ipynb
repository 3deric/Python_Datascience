```python
from numpy.random import randn
```

```python
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5,4), index = 'A B C D E'.split(), columns = 'W X Y Z'.split())
```

```python
df
```

```python
df > 0
#returns true if the cell value is greater then 0
```

```python
df[df > 0]
#return only values that are greater 0, otherwise return NaN
```

```python
df['W'] > 0
#only return true or false for a specific colum
```

```python
df[df['W'] > 0]
#returns the dataframe but ONLY where the values in column 'W' is greather then zero in the rows.
```

```python
df[df['W'] > 0]['Y']
#same as previous, but it only displays column 'Y'
```

```python
df[df['W'] > 0][['X','Y']]
#output of multiple columns where 'W' is greather then zero for the roews
```
