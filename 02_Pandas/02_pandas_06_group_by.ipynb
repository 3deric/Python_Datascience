```python
import pandas as pd

data = {'Company': ['GOOG', 'GOOG', 'MSFT', 'MSFT', 'FB', 'FB'], 
        'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Anne', 'Karl'], 
        'Sales': [200,120,340,124,243,350]}
#create data
```

```python
df = pd.DataFrame(data)
#create dataframe from data
```

```python
df
```

```python
df.groupby('Company')['Sales'].mean()
#get the mean value of each group. Grouped by company. 
#The requested group value needs to be passed too. 
#In this case it is 'Sales'
```

```python
by_company = df.groupby('Company')
#group by company
```

```python
by_company['Sales'].std()
#standard deviation per group
```

```python
by_company['Sales'].min()
#min per company
```

```python
by_company['Sales'].count()
#count the amount of sales per company
```

```python
by_company.describe()
#retrieve information by group
```

```python
by_company.describe().transpose()
#transpose the matrix
```

```python
by_company.describe().transpose()['GOOG']
```
