```python
import numpy as np
import pandas as pd
from numpy.random import randn
np.random.seed(101)
```

```python
outer = ['G1', 'G1', 'G1', 'G2', 'G2', 'G2']
inner = [1, 2, 3, 1, 2, 3]
#creation of outer and inner index
h_index = list(zip(outer, inner))
h_index = pd.MultiIndex.from_tuples(h_index)
#hierarchical index
```

```python
h_index
```

```python
df = pd.DataFrame(np.random.randn(6,2),index = h_index, columns = ['A', 'B'])
#generate a daraframe with multiindex
```

```python
df
```

```python
df.loc['G1']
#output only the outer index 'G1'
```

```python
df.loc['G1'].loc[2]
#access the inner index  2 0f 'G1'
```

```python
df.index.names = ['Group', 'Num']
#set name for the outer indices
```

```python
df
```

```python
df.xs('G1')
#cross section of a index
```

```python
df.xs(('G1',1))
#get Num 1 of Group G1
```

```python
df.xs(1, level = 'Num')
#return all outer index values where the Num index is 1
```
