```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
# allow inline priting of matplotlib output in jupyter
```

```python
x = np.linspace(0,5,11)
y = x**2
# create a simple dataset with 10 points
```

```python
plt.plot(x, y, 'r')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Squared X value')
# plt.show() only required if matplotlib is used outside of jupyter
# plot the graph with the colour red
```

```python
plt.subplot(1,2,1)
plt.plot(x, y, 'r--')
# output two subplots, the left one has a red line
plt.subplot(1,2,2)
plt.plot(y, x, 'g*-')
# output the second plot with a dotted green line
```

```python
fig = plt.figure()

axes = fig.add_axes([0.1,0.1,0.8,0.5])

axes.plot(x, y, 'b')
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title('Beautiful Graph')

# object oriented approach, allows to define the size of the plot
```

```python
af = plt.figure()

axes1 = af.add_axes([0.1,0.1,0.8,0.8])
axes2 = af.add_axes([0.2,0.5,0.4,0.3])

axes1.plot(x,y, 'b')
axes1.set_xlabel('X Label')
axes1.set_ylabel('Y Label')
axes1.set_title('Axes 1 Title')

axes2.plot(y,x, 'g')
axes2.set_xlabel('X Label')
axes2.set_ylabel('Y Label')
axes2.set_title('Axes 2 Title')
```

```python
diag, axes = plt.subplots(nrows = 1, ncols = 3)

for ax in axes:
    ax.plot(x, y)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_title('Title')
#create multiple graphs in a loop

plt.tight_layout()
# remove overlaps
```

```python
diag = plt.figure(figsize = (8,4), dpi = 100)
ax = diag.add_axes([0.0,0.0, 1.0, 1.0])
ax.plot(x, y)
```

```python
diag, axes = plt.subplots(figsize = (8, 3), dpi = 100)
axes.plot(x, y, 'g*-')
```

```python
diag = plt.figure()

ax = diag.add_axes([0.0, 0.0, 1.0, 1.0])
ax.plot(x, x**2, label = 'X Squared')
ax.plot(x, x**3, label = 'X Cubic')

ax.legend(loc = 2)
# loc defines the position between 0 and 10, 0 is best

# multiple plots with a legend
```
