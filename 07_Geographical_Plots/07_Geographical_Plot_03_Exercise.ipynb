## Plotly Importe

```python
import plotly.graph_objs as go 
from plotly.offline import init_notebook_mode,iplot,plot
init_notebook_mode(connected=True) 
```

### Aufgabe: Energienutzung
**Importiere Pandas und lese die CSV-Daten: 2014_World_Power_Consumption**

```python
import pandas as pd
df = pd.read_csv('./examples/2014_World_Power_Consumption')
```

```python
df.head()
```

**Kontrolliere die ersten Einträge im DataFrame.**

**Unter Rückgriff auf die Notizen der Lektion: Erstelle ein Flächendiagramm der Energienutzung (en. power consumption) der Länder durch Verwendung von Daten- und Layout-Dictionary.**

```python
data = dict(type = 'choropleth',
            locations = df['Country'],
            locationmode = 'country names',
            colorscale = 'Portland',
            z = df['Power Consumption KWH'],
            text = df['Text'],
            marker = dict(line = dict(color = 'rgb(0,255,255)', width = 2)),
            colorbar = {'title': 'kWh'})
```

```python
layout = dict(title = 'Country Power Consumption',
              geo = dict(showframe = False,
                         projection={'type': 'natural earth1'}))
```

```python
choromap = go.Figure(data = [data], layout = layout)
```

```python
iplot(choromap)
```

### Aufgabe: Wahlen USA
**Importiere die 2012_Election_Data CSV-Datei indem du Pandas nutzt.**

```python
df = pd.read_csv('./examples/2012_Election_Data')
```

**Kontrolliere die ersten Einträge im neuen DataFrame.**

```python
df.info()
```

```python
df.head()
```

**Erstelle jetzt eine Flächenkarte, die die wahlberechtigte Bevölkerung (en. voting-age population (VAP)) pro Staat zeigt.**

```python
data = dict(type = 'choropleth',
            locations = df['State Abv'],
            locationmode = 'USA-states',
            colorscale = 'Portland',
            z = df['Voting-Age Population (VAP)'],
            text = df['Voting-Age Population (VAP)'],
            marker = dict(line = dict(color = 'rgb(0,255,255)', width = 2)),
            colorbar = {'title': 'VAP'})
```

```python
layout = dict(title = 'Voting-Age Population (VAP)', geo = {'scope': 'usa'})
```

```python
choromap = go.Figure(data = [data], layout = layout)
```

```python
iplot(choromap)
```
