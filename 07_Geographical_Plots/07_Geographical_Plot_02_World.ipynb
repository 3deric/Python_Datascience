```python
import pandas as pd
```

```python
import plotly as py
import plotly.graph_objs as go
from plotly.offline import download_plotlyjs, init_notebook_mode, plot,iplot
```

```python
init_notebook_mode(connected=True)
```

```python
df = pd.read_csv('./examples/2014_World_GDP')
```

```python
df.head()
```

```python
data = dict(type = 'choropleth',
            locations = df['CODE'],
            locationmode = 'ISO-3',
            colorscale = 'Portland',
            z = df['GDP (BILLIONS)'],
            text = df['COUNTRY'],
            marker = dict(line = dict(color = 'rgb(0,255,255)', width = 2)),
            colorbar = {'title': 'GDP Billion USD'})
```

```python
layout = dict(title = '2014 GDP',
              geo = dict(showframe = False,
                         projection={'type': 'orthographic'}))
```

```python
choromap = go.Figure(data = [data], layout = layout)
```

```python
iplot(choromap)
```
