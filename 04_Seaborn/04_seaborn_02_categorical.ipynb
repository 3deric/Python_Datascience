```python
import seaborn as sns
import numpy as np
%matplotlib inline
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.barplot(x = 'sex', y = 'total_bill', data = tips, estimator = 'std')
# show standard deviation of the total bill based on visitor sex
```

```python
sns.countplot(x = 'sex', data = tips)
# show count of visitors
```

```python
sns.boxplot(data = tips, palette = 'rainbow', orient = 'h')
```

```python
sns.boxplot(x = 'day', y = 'total_bill', palette = 'coolwarm', hue = 'smoker', data = tips)
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = tips, palette = 'Set1', hue = 'sex')
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = tips, palette = 'Set1', hue = 'sex', split = True)
# overlap the violin plots
```

```python
sns.stripplot(x = 'day', y = 'total_bill', data = tips, palette = 'Set1', hue = 'sex', jitter = True, dodge = True)
# Jitter separates the points
# Dodge moves the legend away from the points
```

```python
sns.swarmplot(x = 'day', y = 'total_bill', data = tips, palette = 'rainbow', hue = 'sex', dodge = True)
```

```python
sns.violinplot(x = 'tip', y = 'day', data = tips, palette = 'rainbow', hue = 'sex', split = True)
sns.swarmplot(x = 'tip', y = 'day', data = tips, size = 3, color = 'black')
# combined violin and swarmplot
```

```python
sns.catplot(x = 'sex', y = 'total_bill', data = tips, kind = 'bar', palette = 'Set1', hue = 'sex')
```
