```python
import seaborn as sns
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'coolwarm',
          markers = ['o','v'], scatter_kws = {'s' : 20})
# markers are passed as a array, circle for male, triangle for female
# size of markers is defined in a dictionary, parameter s defines the size
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, col = 'day', hue = 'sex', aspect = 0.6)
```
