```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
plt.figure(figsize = (10,5))
# set figure size of the seaborn plot
sns.set_style('ticks')
sns.countplot(x = 'sex', data = tips, palette = 'deep', hue = 'sex')
sns.despine()
# remove right and upper frame
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', height = 2, aspect = 4, data = tips)
```

```python
sns.set_context('talk', font_scale = 1)
sns.countplot(x = 'sex', data = tips, palette = 'magma', hue = 'sex')
sns.despine(left = True, bottom = True)
# remove all outlines
```
