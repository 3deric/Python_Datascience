**Kernel Distribution Plot without Seaborn**

```python
# manual creation of kernel distribution without seaborn native functions

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

dataset = np.random.randn(25)
# create dataset

sns.rugplot(dataset)
# create another rugplot

x_min = dataset.min() -2
x_max = dataset.max() +2
# create the plot axis

x_axis = np.linspace(x_min, x_max, 100)
# create 100 evenly distributed points between min and max

bandwidth = ((4* dataset.std() ** 5)/(3*len(dataset)))**0.2
# create bandwidth

kernel_list = []
# create a empty kernel list

for data_point in dataset:
    kernel = stats.norm(data_point, bandwidth).pdf(x_axis)
    kernel_list.append(kernel)

    kernel = kernel / kernel.max()
    kernel = kernel * 0.4
    plt.plot(x_axis, kernel, color = 'grey', alpha = 0.5)
plt.ylim(0,1)
```

```python
sum_of_kde = np.sum(kernel_list, axis = 0)

fig = plt.plot(x_axis, sum_of_kde, color = 'indianred')

sns.rugplot(dataset, c = 'indianred')

plt.yticks([])

plt.suptitle('Sum of Basefunctions')
```
